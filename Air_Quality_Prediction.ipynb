**Handling the missing values**

```python
# convert all -200 to NaN
air_quality_data =air_quality_data.replace(to_replace = -200, value = np.nan)
```

```python
air_quality_data.tail()
```

```python
# Calculate the mean of numeric columns, excluding 'Date' and 'Time'
numeric_columns = air_quality_data.select_dtypes(include=np.number)
numeric_columns.mean()
```

```python
# replace all the NaN to the mean of the particular columN
air_quality_data = air_quality_data.fillna(numeric_columns.mean())
```

```python
air_quality_data.tail()
```

**Forecasting with FB Prophet Algorithm**

FB Prophet Documentation : [https://facebook.github.io/prophet/docs/quick_start.html](https://facebook.github.io/prophet/docs/quick_start.html)

```python
# convert date from DD/MM/YYYY TO YYYY/MM/DD
date_info= pd.to_datetime(air_quality_data['Date'], format='%d/%m/%Y')
print(date_info)
```

```python
time_info= (air_quality_data['Time'])
print(time_info)
```

```python
time_info = time_info.apply(lambda x: x.replace('.',':'))
print(time_info)
```

```python
print(type(date_info))
print(type(time_info))
```

```python
# combining two series to pandas dataframe
date_time_info = pd.concat([date_info,time_info],axis=1)
date_time_info.head()
```

```python
date_time_info.shape
```

```python
# combining date and time
date_time_info['ds'] = date_time_info['Date'].astype(str) + ' ' + date_time_info['Time'].astype(str)
date_time_info.head()
```

```python
date_time_info.info()
```

We have to convert ds from object datatype to date time format.

```python
data = pd.DataFrame()
```

```python
data['ds'] = pd.to_datetime(date_time_info['ds'])
data.info()
```

```python
data.head()
```

```python
data['y']=air_quality_data['RH']
```

```python
data.head()
```

```python
model = Prophet()
model.fit(data)
```

```python
future = model.make_future_dataframe(periods=365,freq='H')
future.tail()
```

```python
forecast = model.predict(future)
forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail()
```

```python
fig1 = model.plot(forecast)
```

```python
fig2 = model.plot_components(forecast)
```
